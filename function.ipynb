4.What is the purpose of the `return` statement in a Python function?
 - The return statement in Python is used to exit a function and send a value back to the place where the function was called. It allows the function to produce output that can be stored or used elsewhere in the program. If no return is specified, the function returns None by default.

5.What are iterators in Python and how do they differ from iterables?
 - In Python, an iterator is an object that represents a stream of data and returns one element at a time using the next() function until all items are exhausted. It must implement two methods: __iter__() and __next__().
An iterable is any Python object capable of returning its elements one at a time, such as lists, tuples, or strings. It only needs to implement the __iter__() method and returns an iterator when iter() is called on it.

6.Explain the concept of generators in Python and how they are defined.
 - In Python, generators are a special type of iterator that allow you to iterate over a sequence of values lazily, meaning they generate items on the fly and do not store them in memory.
 - Defined using a regular function but with the **yield** keyword instead of return.

7.What are the advantages of using generators over regular functions?
 - Generators yield items one at a time, so they don't store the entire result in memory like lists do.
 - Values are produced only when needed, reducing unnecessary computation.
 - Generators can simplify code that would otherwise require complex state tracking.

9.Explain the purpose and usage of the `map()` function in Python.
 - Purpose:
   - Transform data without using explicit loops.
   - Can be more concise and functional compared to traditional for loops.
 - usage:
   - Apply a transformation to each element of an iterable (e.g., squaring numbers, converting strings to uppercase).
   - You can pass multiple iterables to map(), and the function should accept as many arguments as there are iterables.

10.What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
 - map()
   - Purpose: Applies a given function to each item of an iterable (or multiple iterables) and returns an iterator of the results.
   - Usage: Used to transform elements of an iterable.
 - filter()
   - Purpose: Filters elements of an iterable based on a function that returns either True or False. Only elements for which the function returns True are included in the result.
   - Usage: Used to filter out elements that don't meet a certain condition.
 - reduce()
   - Purpose: Applies a function of two arguments cumulatively to the items of an iterable (from left to right) so as to reduce the iterable to a single value.
  - Usage: Used to combine all elements of an iterable into one value, such as summing numbers or multiplying elements.

```python
# Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.
def sum_of_evens(numbers):
    return sum(num for num in numbers if num % 2 == 0)
numbers = [1, 2, 3, 4, 5, 6]
result = sum_of_evens(numbers)
print(result)
```

```python
# Implement a Python function that takes a list of integers and returns a new list containing the squares of each number
def square_numbers(numbers):
    return [num ** 2 for num in numbers]
numbers = [1, 2, 3, 4, 5]
result = square_numbers(numbers)
print(result)
```

```python
# Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
for num in range(1, 201):
    if is_prime(num):
        print(num)
```

```python
#Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class FibonacciIterator:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        fib = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return fib
n_terms = 10
fib_iterator = FibonacciIterator(n_terms)

for num in fib_iterator:
    print(num)
```

```python
#Write a generator function in Python that yields the powers of 2 up to a given exponent.
def powers_of_2(exponent):
    for i in range(exponent + 1):
        yield 2 ** i
exponent = 5
for power in powers_of_2(exponent):
    print(power)
```

```python
#Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()
filename = "example.txt"
for line in read_file_line_by_line(filename):
    print(line)
```

```python
# Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.
tuples = [(1, 5), (3, 2), (4, 8), (2, 3)]
sorted_tuples = sorted(tuples, key=lambda x: x[1])
print(sorted_tuples)
```

```python
#Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
celsius_temperatures = [0, 25, 30, 35, 40]
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
fahrenheit_temperatures = list(map(celsius_to_fahrenheit, celsius_temperatures))
print(fahrenheit_temperatures)
```

```python
# Create a Python program that uses `filter()` to remove all the vowels from a given string.
def is_not_vowel(char):
    return char.lower() not in 'aeiou'
input_string = "Hello, World!"
filtered_string = ''.join(filter(is_not_vowel, input_string))
print(filtered_string)
```
